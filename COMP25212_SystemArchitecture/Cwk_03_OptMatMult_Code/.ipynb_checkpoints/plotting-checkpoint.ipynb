# Baseline

# Compiler optimisation

```python
O_df = pd.read_csv("O_data.txt", header = None)
O_df.columns = ["L", "ExecTime"]

O1_df = pd.read_csv("O1_data.txt", header = None)
O1_df.columns = ["L", "ExecTime"]

O2_df = pd.read_csv("O2_data.txt", header = None)
O2_df.columns = ["L", "ExecTime"]

O3_df = pd.read_csv("O3_data.txt", header = None)
O3_df.columns = ["L", "ExecTime"]

O_df["FLOP"] = O_df["L"] * O_df["L"] * (2 * O_df["L"] - 1)
O_df["FLOPS"] = O_df["FLOP"] / O_df["ExecTime"]

O1_df["FLOP"] = O1_df["L"] * O1_df["L"] * (2 * O1_df["L"] - 1)
O1_df["FLOPS"] = O1_df["FLOP"] / O1_df["ExecTime"]

O2_df["FLOP"] = O2_df["L"] * O2_df["L"] * (2 * O2_df["L"] - 1)
O2_df["FLOPS"] = O2_df["FLOP"] / O2_df["ExecTime"]

O3_df["FLOP"] = O3_df["L"] * O3_df["L"] * (2 * O3_df["L"] - 1)
O3_df["FLOPS"] = O3_df["FLOP"] / O3_df["ExecTime"]
```

# Loop unrolling

```python
un2_df = pd.read_csv("unrolling2_data.txt", header = None)
un2_df.columns = ["L", "ExecTime"]

un4_df = pd.read_csv("unrolling4_data.txt", header = None)
un4_df.columns = ["L", "ExecTime"]

un8_df = pd.read_csv("unrolling8_data.txt", header = None)
un8_df.columns = ["L", "ExecTime"]

un16_df = pd.read_csv("unrolling16_data.txt", header = None)
un16_df.columns = ["L", "ExecTime"]

un32_df = pd.read_csv("unrolling32_data.txt", header = None)
un32_df.columns = ["L", "ExecTime"]

un2_df["FLOP"] = un2_df["L"] * un2_df["L"] * (2 * un2_df["L"] - 1)
un2_df["FLOPS"] = un2_df["FLOP"] / un2_df["ExecTime"]

un4_df["FLOP"] = un4_df["L"] * un4_df["L"] * (2 * un4_df["L"] - 1)
un4_df["FLOPS"] = un4_df["FLOP"] / un4_df["ExecTime"]

un8_df["FLOP"] = un8_df["L"] * un8_df["L"] * (2 * un8_df["L"] - 1)
un8_df["FLOPS"] = un8_df["FLOP"] / un8_df["ExecTime"]

un16_df["FLOP"] = un16_df["L"] * un16_df["L"] * (2 * un16_df["L"] - 1)
un16_df["FLOPS"] = un16_df["FLOP"] / un16_df["ExecTime"]

un32_df["FLOP"] = un32_df["L"] * un32_df["L"] * (2 * un32_df["L"] - 1)
un32_df["FLOPS"] = un32_df["FLOP"] / un32_df["ExecTime"]
```

# Blocking

```python
block2_df = pd.read_csv("block2_data.txt", header = None)
block2_df.columns = ["L", "ExecTime"]

block4_df = pd.read_csv("block4_data.txt", header = None)
block4_df.columns = ["L", "ExecTime"]

block8_df = pd.read_csv("block8_data.txt", header = None)
block8_df.columns = ["L", "ExecTime"]

block16_df = pd.read_csv("block16_data.txt", header = None)
block16_df.columns = ["L", "ExecTime"]

block32_df = pd.read_csv("block32_data.txt", header = None)
block32_df.columns = ["L", "ExecTime"]

block64_df = pd.read_csv("block64_data.txt", header = None)
block64_df.columns = ["L", "ExecTime"]

block128_df = pd.read_csv("block128_data.txt", header = None)
block128_df.columns = ["L", "ExecTime"]

block2_df["FLOP"] = block2_df["L"] * block2_df["L"] * (2 * block2_df["L"] - 1)
block2_df["FLOPS"] = block2_df["FLOP"] / block2_df["ExecTime"]

block4_df["FLOP"] = block4_df["L"] * block4_df["L"] * (2 * block4_df["L"] - 1)
block4_df["FLOPS"] = block4_df["FLOP"] / block4_df["ExecTime"]

block8_df["FLOP"] = block8_df["L"] * block8_df["L"] * (2 * block8_df["L"] - 1)
block8_df["FLOPS"] = block8_df["FLOP"] / block8_df["ExecTime"]

block16_df["FLOP"] = block16_df["L"] * block16_df["L"] * (2 * block16_df["L"] - 1)
block16_df["FLOPS"] = block16_df["FLOP"] / block16_df["ExecTime"]

block32_df["FLOP"] = block32_df["L"] * block32_df["L"] * (2 * block32_df["L"] - 1)
block32_df["FLOPS"] = block32_df["FLOP"] / block32_df["ExecTime"]

block64_df["FLOP"] = block64_df["L"] * block64_df["L"] * (2 * block64_df["L"] - 1)
block64_df["FLOPS"] = block64_df["FLOP"] / block64_df["ExecTime"]

block128_df["FLOP"] = block128_df["L"] * block128_df["L"] * (2 * block128_df["L"] - 1)
block128_df["FLOPS"] = block128_df["FLOP"] / block128_df["ExecTime"]
```

# Combination

### Choosing the right combination

In the first stage, the different parameters are kept constant. This first stage allows for different methods to be explored before the "best" one is chosen.

```python
comb1_df = pd.read_csv("combi1_data.txt", header = None)
comb1_df.columns = ["L", "ExecTime"]

comb1_df["FLOP"] = comb1_df["L"] * comb1_df["L"] * (2 * comb1_df["L"] - 1)
comb1_df["FLOPS"] = comb1_df["FLOP"] / comb1_df["ExecTime"]

comb2_df = pd.read_csv("combi2_data.txt", header = None)
comb2_df.columns = ["L", "ExecTime"]

comb2_df["FLOP"] = comb2_df["L"] * comb2_df["L"] * (2 * comb2_df["L"] - 1)
comb2_df["FLOPS"] = comb2_df["FLOP"] / comb2_df["ExecTime"]

comb3_df = pd.read_csv("combi3_data.txt", header = None)
comb3_df.columns = ["L", "ExecTime"]

comb3_df["FLOP"] = comb3_df["L"] * comb3_df["L"] * (2 * comb3_df["L"] - 1)
comb3_df["FLOPS"] = comb3_df["FLOP"] / comb3_df["ExecTime"]

comb4_df = pd.read_csv("combi4_data.txt", header = None)
comb4_df.columns = ["L", "ExecTime"]

comb4_df["FLOP"] = comb4_df["L"] * comb4_df["L"] * (2 * comb4_df["L"] - 1)
comb4_df["FLOPS"] = comb4_df["FLOP"] / comb4_df["ExecTime"]

comb5_df = pd.read_csv("combi5_data.txt", header = None)
comb5_df.columns = ["L", "ExecTime"]

comb5_df["FLOP"] = comb5_df["L"] * comb5_df["L"] * (2 * comb5_df["L"] - 1)
comb5_df["FLOPS"] = comb5_df["FLOP"] / comb5_df["ExecTime"]

comb6_df = pd.read_csv("combi6_data.txt", header = None)
comb6_df.columns = ["L", "ExecTime"]

comb6_df["FLOP"] = comb6_df["L"] * comb6_df["L"] * (2 * comb6_df["L"] - 1)
comb6_df["FLOPS"] = comb6_df["FLOP"] / comb6_df["ExecTime"]
```

```python
plt.figure(figsize = (12, 6))
plt.plot(comb1_df["L"], comb1_df["FLOPS"], label = "Multicore + blocking")
plt.plot(comb2_df["L"], comb2_df["FLOPS"], label = "Multicore + blocking + SIMD")
plt.plot(comb3_df["L"], comb3_df["FLOPS"], label = "Multicore + blocking + SIMD + compiler unrolling")
plt.plot(comb4_df["L"], comb4_df["FLOPS"], label = "Multicore + blocking + SIMD + compiler unrolling + prefetch")
plt.plot(comb5_df["L"], comb5_df["FLOPS"], label = "Multicore + blocking + compiler unrolling")
plt.plot(comb6_df["L"], comb6_df["FLOPS"], label = "Multicore + blocking + compiler unrolling + prefetch")
plt.plot(O3_df["L"], O3_df["FLOPS"], label = "-O3 optimisation")
plt.ylabel("FLOPS")
plt.xlabel("Matrix size")
plt.title("Combination of methods - FLOPS vs matrix size")
plt.legend(loc = 2)
plt.grid()
plt.savefig("combination.jpg")
```
