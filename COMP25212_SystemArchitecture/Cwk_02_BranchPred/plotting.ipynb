```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = np.loadtxt("data_collection.txt")
df2 = np.loadtxt("data_hysteresis.txt")
df2
```

```python
index = [x for x in range(1, 201)]

plt.figure(figsize = (10, 6))
plt.plot(index, df, label = "Without hysteresis")
plt.plot(index, df2, label = "With hysteresis")
plt.grid()
plt.xlabel("BTB size")
plt.ylabel("Instructions fetched / instructions executed (%)")
plt.legend()
plt.savefig("overhead.png")
```
