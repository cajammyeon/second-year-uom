```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('test_data.csv', header = None)
df.head(20)
```

```python
gradient, intercept = np.polyfit(df[0], df[1], 1)
print(f"gradient : {gradient} \nintercept : {intercept}")

plt.figure(figsize=(15,9))
plt.plot(df[0], df[1])
plt.plot(df[0], df[0]*gradient+intercept)

plt.xlabel("Capacity of knapsack W")
plt.ylabel("Time taken for the algorithm to compute solution T (ns)")
plt.title("T against W")
plt.grid()
plt.savefig("graph.png")
```
